Configurar Hadoop en modo pseudo-distribuido:

```python
import os

HADOOP_HOME = "/content/hadoop-3.4.2"
config_dir = f"{HADOOP_HOME}/etc/hadoop"

# Configurar el archivo core-site.xml
core_site = """<?xml version="1.0"?>
<configuration>
    <property>
        <name>fs.defaultFS</name>
        <value>hdfs://localhost:9000</value>
    </property>
</configuration>
"""

with open(f"{config_dir}/core-site.xml", "w") as f:
    f.write(core_site)

# Configurar el archivo hdfs-site.xml
hdfs_site = """<?xml version="1.0"?>
<configuration>
    <property>
        <name>dfs.replication</name>
        <value>1</value>
    </property>
    <property>
        <name>dfs.namenode.name.dir</name>
        <value>file:///content/hadoopdata/namenode</value>
    </property>
    <property>
        <name>dfs.datanode.data.dir</name>
        <value>file:///content/hadoopdata/datanode</value>
    </property>
</configuration>
"""

with open(f"{config_dir}/hdfs-site.xml", "w") as f:
    f.write(hdfs_site)

# Crear directorios de almacenamiento
os.makedirs("/content/hadoopdata/namenode", exist_ok=True)
os.makedirs("/content/hadoopdata/datanode", exist_ok=True)

print("Archivos de configuración creados correctamente.")
```

```python
%%writefile mapper.py

import sys
import csv

reader = csv.reader(sys.stdin, delimiter=';')

for row in reader:
    if not row:
        continue

    if row[-1].strip() == "Target":
        continue

    target = row[-1].strip()

    if target:
        print(f"{target}\t1")
```

Crear el Reducer:

```python
%%writefile reducer.py

import sys

current_key = None
current_count = 0

for line in sys.stdin:
    key, value = line.strip().split("\t")

    if current_key == key:
        current_count += int(value)
    else:
        if current_key is not None:
            print(f"{current_key}\t{current_count}")

        current_key = key
        current_count = int(value)

if current_key is not None:
    print(f"{current_key}\t{current_count}")
```

Ejecutar Map Reduce:

```python
!hdfs dfs -rm -r -f /user/root/bigdata/output_target

!hadoop jar /content/hadoop-3.4.2/share/hadoop/tools/lib/hadoop-streaming-3.4.2.jar \
    -D mapreduce.framework.name=local \
    -input /user/root/bigdata/input/data.csv \
    -output /user/root/bigdata/output_target \
    -mapper "python3 mapper.py" \
    -reducer "python3 reducer.py" \
    -file mapper.py \
    -file reducer.py
```
